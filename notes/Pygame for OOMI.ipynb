

## 🎮 Pygame Lesson 1 – Window with Custom Background Image


### 1. Theory (read carefully)

A pygame program needs:
- **Initialization** (`pygame.init()`)
- **Window** (`pygame.display.set_mode()`)
- **Background image** – loaded from file, scaled to fit window, then drawn each frame
- **Event loop** – to keep window open and respond to closing

**Key points about background image:**
- Load image with `pygame.image.load("filename")`
- Scale to window size with `pygame.transform.scale()`
- Draw it each frame with `screen.blit(bg_image, (0,0))` **before** any other drawings (cards, text)
- If the image file is missing, fallback to a solid color so the program doesn't crash

**Minimal code with background:**

```python
import pygame
import sys

pygame.init()
WIDTH, HEIGHT = 1024, 768
screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Omi – Sri Lankan Card Game")

# Load background image (put your image file in the same folder)
try:
    bg_image = pygame.image.load("sri_lankan_bg.jpg")
    bg_image = pygame.transform.scale(bg_image, (WIDTH, HEIGHT))
except FileNotFoundError:
    # Fallback: create a solid color background
    bg_image = None
    print("Background image not found, using solid color.")

# Game loop
while True:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            pygame.quit()
            sys.exit()

    # Draw background
    if bg_image:
        screen.blit(bg_image, (0, 0))
    else:
        screen.fill((50, 70, 30))  # dark earthy green

    pygame.display.flip()
```

# 🎮 Pygame Lesson 2 – Drawing Cards (Rectangles + Text)





### 1. Theory

In pygame, you draw shapes and text every frame. For a card:

- **Rectangle:** `pygame.draw.rect(screen, color, (x, y, width, height))`
- **Text:** create a `Font`, `render()` the text into an image, then `blit()` it onto the screen.

**Card size:** We'll use 70x90 pixels.  
**Colors:** White for card face, black for text, red for hearts/diamonds, black for spades/clubs.

**Font:** Use `pygame.font.SysFont(None, 24)` for a default font size 24.

```python
import pygame
import sys

pygame.init()
WIDTH, HEIGHT = 800, 600
screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Card Test")

font = pygame.font.SysFont(None, 24)

# Card position and size
x, y = 100, 400
w, h = 70, 90

while True:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            pygame.quit()
            sys.exit()

    screen.fill((50, 70, 30))  # dark green background

    # Draw card background (white)
    pygame.draw.rect(screen, (255,255,255), (x, y, w, h), border_radius=5)
    # Draw border (black)
    pygame.draw.rect(screen, (0,0,0), (x, y, w, h), 2, border_radius=5)

    # Draw text (Ace of Spades)
    text_surface = font.render("A♠", True, (0,0,0))
    screen.blit(text_surface, (x + 10, y + 30))

    pygame.display.flip()
```

```python
import pygame
import sys

pygame.init()
WIDTH, HEIGHT = 800, 600
screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Face‑down Card Test")

CARD_W, CARD_H = 70, 90
font = pygame.font.SysFont(None, 24)

def draw_face_down(x, y):
    # Draw grey rectangle
    pygame.draw.rect(screen, (100,100,100), (x, y, CARD_W, CARD_H), border_radius=5)
    pygame.draw.rect(screen, (0,0,0), (x, y, CARD_W, CARD_H), 2, border_radius=5)
    # Draw "?"
    text = font.render("?", True, (255,255,255))
    screen.blit(text, (x + CARD_W//2 - 10, y + CARD_H//2 - 10))

running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False
    screen.fill((50,70,30))
    draw_face_down(100, 200)
    pygame.display.flip()
pygame.quit()
```

# 📝 Note: Adding Mouse Click Detection (Human Card Selection)




### 1. Theory

In pygame, you can detect mouse clicks by checking `event.type == pygame.MOUSEBUTTONDOWN` inside the event loop. Then you get the mouse position with `pygame.mouse.get_pos()`. To see if a card was clicked, you need to calculate the rectangle (top‑left corner and width/height) of each card in the human hand and test if the mouse point is inside that rectangle.

**Key points:**
- The human cards are drawn using `draw_card(x + CARD_W//2, human_y, ...)`, where `x = human_x_start + i * (CARD_W + 10)`.
- The **top‑left corner** of the card is `(x, human_y - CARD_H//2)`.
- The card rectangle is `(x, human_y - CARD_H//2, CARD_W, CARD_H)`.
- Store the selected card index, and draw a highlight (e.g., a yellow border) around it.

### 2. Test – Write a small script

Create a new file `click_test.py`. Draw a single rectangle (like a card) and print when clicked. Example:

```python
import pygame
pygame.init()
screen = pygame.display.set_mode((800,600))
rect = pygame.Rect(200, 200, 70, 90)
while True:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            pygame.quit(); exit()
        if event.type == pygame.MOUSEBUTTONDOWN:
            if rect.collidepoint(event.pos):
                print("Card clicked!")
    screen.fill((0,0,0))
    pygame.draw.rect(screen, (255,255,255), rect)
    pygame.display.flip()
```

### 1. Where to put the code

Inside your main game loop, after `for event in pygame.event.get():` and after the `QUIT` check, add a new condition:

```python
if event.type == pygame.MOUSEBUTTONDOWN:
    # get mouse position
    mouse_x, mouse_y = event.pos
    # loop through human cards to see if any was clicked
    for i, card in enumerate(human_hand):
        # compute top-left corner of this card
        card_x = human_x_start + i * (CARD_W + 10)
        card_top = human_y - CARD_H // 2   # because card is drawn centered at human_y
        # create a rectangle
        card_rect = pygame.Rect(card_x, card_top, CARD_W, CARD_H)
        if card_rect.collidepoint(mouse_x, mouse_y):
            selected_card_index = i
            # you can also print which card was clicked
            print(f"Clicked card {i}: {card}")
```
